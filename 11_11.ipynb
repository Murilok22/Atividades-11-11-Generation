```python
carros = ["Uno", "Corola", "Fusca"]

# para usar o for com mais de uma lista elas devem ter o mesmo tamanho
for i in carros:
    print(f"{i}")
print("")


# Cadastro de novos produtos
resposta = input("Deseja cadastrar um novo carro? ")
while resposta.lower() == "sim":
    novo_carros = input("Digite o nome do novo carro: ")
    #anexando novo produto
    carros.append(novo_carros)
    resposta = input("Deseja parar o cadastro de novos veiculos? ")
    if resposta.lower() != "sim":
      break


# Lista atual, apos inserir
print("\n====== Produtos Atuais ======")
for i in carros:
    print(f"{i}")
print("")


# Exclusão de produtos, conferindo se existem itens na lista
if len(carros) > 0:
    resposta = input("Deseja excluir o ultimo carro ? ")
    if resposta.lower() == "sim":
            carros.pop()
            print(f"Modelo {i} excluido")
    else:
      print("Nenhum produto cadastrado para exclusão.")



# Lista atual, apos pop
print("\n====== Produtos Atuais ======")
for i in carros:
    print(f"{i}")
print("\n")
```
